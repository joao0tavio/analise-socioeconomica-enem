```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('dados/enem_2021_limpo.csv')

# Correlação: O primeiro "teste de sanidade"
# A correlação vai de -1 a 1. Quanto mais próximo de 1, mais forte a relação.
print("Matriz de Correlação:")
print(df[['RENDA_NUMERICA', 'NOTA_FINAL']].corr())

# Visualização 1: O Boxplot
# Ele mostra a distribuição das notas para CADA classe de renda.
# Se a hipótese for real, as "caixas" devem subir conforme a renda sobe.
ordem_renda = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J', 'K', 'L', 'M', 'N', 'O', 'P', 'Q']
plt.figure(figsize=(12, 6))
sns.boxplot(x='Q006', y='NOTA_FINAL', data=df, order=ordem_renda)
plt.title('Distribuição da Nota Final por Faixa de Renda')
plt.xlabel('Renda (Classes A -> Q)')
plt.ylabel('Nota Final')
plt.show()

# Visualização 2: Gráfico de Dispersão (Scatterplot) resumido
# Como são 2 milhões de pontos, o gráfico fica pesado. Vamos pegar uma amostra aleatória
# para visualizar a tendência da Regressão Linear.
amostra = df.sample(10000) # Pegando 10 mil alunos aleatórios
plt.figure(figsize=(10, 6))
sns.scatterplot(x='RENDA_NUMERICA', y='NOTA_FINAL', data=amostra, alpha=0.3)
# Adiciona uma linha de tendência vermelha para simular a regressão visualmente
sns.regplot(x='RENDA_NUMERICA', y='NOTA_FINAL', data=amostra, scatter=False, color='red')
plt.title('Renda Numérica vs Nota Final (Amostra)')
plt.show()
```
